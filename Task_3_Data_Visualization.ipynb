```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = sns.load_dataset("titanic")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="age", bins=30, kde=True)

plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="class")

plt.title("Number of Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="sex", hue="survived")

plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="class", y="age")

plt.title("Age Distribution Across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived"
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Titanic Dataset")

plt.show()
```

```python
survival_rate = df.groupby("class")["survived"].mean().reset_index()

plt.figure(figsize=(7, 5))

sns.barplot(
    data=survival_rate,
    x="class",
    y="survived"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

```python
sns.pairplot(
    df[["age", "fare", "sibsp", "parch", "survived"]].dropna(),
    hue="survived"
)

plt.show()
```

## Key Insights

- The age distribution shows that passengers belonged to different age groups, with many passengers concentrated in the younger and middle-age ranges.
- Passenger class had a visible relationship with survival outcomes.
- The survival distribution differed between male and female passengers.
- Passenger fare showed variation across different passenger classes.
- The correlation heatmap helped identify relationships among numerical variables.
- Box plots helped identify the spread and possible outliers in passenger age and fare.
- Scatter plots provided a visual understanding of relationships between numerical features.
