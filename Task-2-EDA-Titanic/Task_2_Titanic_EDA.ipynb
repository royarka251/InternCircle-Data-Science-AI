```python
import pandas as pd
import numpy as np
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = sns.load_dataset("titanic")

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

```python
df.describe(include="all")
```

```python
print("Columns in the dataset:")
print(df.columns.tolist())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Value Percentage:")
print(missing_percentage)
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:")
print(df.shape)
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
df.groupby("sex")["survived"].mean()
```

```python
survival_by_gender = df.groupby("sex")["survived"].agg(["count", "mean"])

survival_by_gender
```

```python
survival_by_class = df.groupby("class")["survived"].agg(["count", "mean"])

survival_by_class
```

```python
df.groupby(["class", "sex"])["survived"].mean()
```

```python
print("Average age:")
print(df["age"].mean())

print("\nMedian age:")
print(df["age"].median())
```

```python
print("Average fare:")
print(df["fare"].mean())

print("\nMaximum fare:")
print(df["fare"].max())

print("\nMinimum fare:")
print(df["fare"].min())
```

## Key Observations

1. The Titanic dataset contains information about passengers, including age, gender, passenger class and survival status.

2. Missing values were identified and handled using appropriate techniques.

3. Duplicate records were checked and removed where necessary.

4. Survival rates were compared across gender and passenger classes.

5. Grouping and aggregation using Pandas helped identify differences in survival patterns.

6. Passenger class and gender can be useful variables for understanding survival outcomes.
