```python
import gradio as gr
import pandas as pd
import joblib

# Load the trained model
model = joblib.load("/content/model.pkl")


# Prediction function
def predict_survival(pclass, sex, age, sibsp, parch, fare, embarked):

    # Create input DataFrame in the same format used during training
    input_data = pd.DataFrame({
        "pclass": [pclass],
        "sex": [sex],
        "age": [age],
        "sibsp": [sibsp],
        "parch": [parch],
        "fare": [fare],
        "embarked": [embarked]
    })

    # Make prediction
    prediction = model.predict(input_data)

    if prediction[0] == 1:
        return "🟢 Prediction: Passenger is likely to survive."
    else:
        return "🔴 Prediction: Passenger is unlikely to survive."


# Create Gradio interface
app = gr.Interface(
    fn=predict_survival,

    inputs=[
        gr.Dropdown(
            choices=[1, 2, 3],
            label="Passenger Class (Pclass)",
            value=3
        ),

        gr.Dropdown(
            choices=["male", "female"],
            label="Sex",
            value="male"
        ),

        gr.Number(
            label="Age",
            value=22
        ),

        gr.Number(
            label="Siblings / Spouses (SibSp)",
            value=0
        ),

        gr.Number(
            label="Parents / Children (Parch)",
            value=0
        ),

        gr.Number(
            label="Fare",
            value=7.25
        ),

        gr.Dropdown(
            choices=["S", "C", "Q"],
            label="Port of Embarkation",
            value="S"
        )
    ],

    outputs=gr.Textbox(
        label="Prediction"
    ),

    title="🚢 Titanic Survival Predictor",

    description="Enter passenger information to predict Titanic survival."
)


print("Gradio interface created successfully!")
```

```python
app.launch(share=True)
```
